```python
import os
```

```python
root = "KhataSetu-Finance"
```

```python
folders = [
    root,
    f"{root}/pages",
    f"{root}/utils",
    f"{root}/data",
]
files = [
        f"{root}/app.py",
    f"{root}/pages/1_🌐_Language_Onboarding.py",
    f"{root}/pages/2_🔐_Data_Consent.py",
    f"{root}/pages/3_🪪_eKYC_Verification.py",
    f"{root}/pages/4_💰_Credit_Offer.py",
    f"{root}/pages/5_✍️_eSign_Disbursement.py",
    f"{root}/pages/6_📊_My_Vyapar_Line.py",
    f"{root}/pages/7_🏦_Lender_Ops_Dashboard.py",
    f"{root}/pages/8_📈_Investor_Metrics.py",
    f"{root}/utils/__init__.py",
    f"{root}/utils/scoring_engine.py",
    f"{root}/utils/synthetic_data.py",
    f"{root}/utils/db.py",
    f"{root}/utils/styling.py",
]

for folder in folders:
    os.makedirs(folder, exist_ok=True)
    print(f"📁 {folder}")

for file in files:
    if not os.path.exists(file):
        open(file, "w").close()
        print(f"📄 {file}")

print("\n✅ Skeleton created.")
```

```python
for dirpath, dirnames, filenames in os.walk("."):
    if ".git" in dirpath or "__pycache__" in dirpath:
        continue
    level = dirpath.count(os.sep)
    indent = "  " * level
    print(f"{indent}{os.path.basename(dirpath)}/")
    for f in filenames:
        print(f"{indent}  {f}")
```
